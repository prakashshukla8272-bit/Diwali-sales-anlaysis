```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("Diwali Sales Data.csv", encoding="latin1")
df
```

```python
df.columns
```

```python
df.columns=df.columns.str.lower()
df.columns
```

```python
                
df.columns = df.columns.str.replace(" ","_")
df.columns
```

```python
df.info()
```

```python
df.shape
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
df  = df.drop(['status' , 'unnamed'],axis =1)
```

```python
df.info()
```

```python
df.dropna(inplace = True)
```

```python
df.isnull().sum()
```

```python
df['amount'] = df['amount'].astype('int')
```

```python
df.info()
```

```python
 df[['amount', 'age', 'orders']].describe()
```

```python
df.columns
```

```python
# Gender
```

```python
ax=sns.countplot(x = 'gender' , data =df)
for bars in ax.containers:
    ax.bar_label(bars)
```

```python
gender1 = df.groupby(['gender'] , as_index =False)['orders'].sum().sort_values(by ='orders' ,ascending = False)
gender1
sns.barplot(x ='gender' , y ='orders' , data  = gender1)
```

```python
    #   AGE
```

```python
ax = sns.countplot(x = 'age_group' , hue = "gender", data = df)

for bars in ax.containers:
     ax.bar_label(bars)
```

```python
age1 =df.groupby(['age_group'] , as_index = False)['amount'].sum().sort_values( by = 'amount' , ascending = False)
```

```python
sns.barplot(x = 'age_group', y = 'amount' ,data = age1)
```

```python
#     STATE
```

```python
df.columns
```

```python
state1  = df.groupby(['state'] , as_index = False)['amount'].sum().sort_values( by  = 'amount' , ascending = False).head(10) 
sns.set(rc= {'figure.figsize' :(15,5)})
sns.barplot( x = "state", y = "amount"  , data =state1 )
    
```

```python
ax= sns.countplot(x= 'marital_status'  ,data =df)
sns.set(rc = {'figure.figsize' : (3,5)})
for bars in ax.containers:
    ax.bar_label(bars)

```

```python
mat1 = df.groupby(['marital_status', 'gender'] , as_index = False )['amount'].sum().sort_values( by = 'amount' , ascending = True)
sns.barplot(x = 'marital_status' , y = 'amount', data = mat1 , hue = 'gender')
sns.set(rc = {'figure.figsize' : (3,5)})

```
